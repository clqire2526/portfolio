```python
%%html

<div id="outputPH1">Results will appear here:<br></div>

<script>
(() => {

    const output = document.getElementById("outputPH1");

    let mood = "happy";

    if (mood === "happy") { 
        console.log("You're happy! You get ice cream.");
        output.innerText += "You're happy! You get ice cream.";
    } else if (mood === "sad") {
        console.log("You're sad. You get a rock.");
        output.innerText += "You're sad. You get a rock.";
    } else if (mood === "okay") {
        console.log("You're okay. You get nothing.");
        output.innerText += "You're okay. You get nothing.";
    }

    let weather = "sunny";

    if (weather === "sunny") {
        console.log("It's sunny outside!");
        output.innerText += "It's sunny outside!";
    } else if (weather === "rainy") {
        console.log("It is raining.");
        output.innerText += "It is raining.";
    } else if (weather === "cloudy") {
        console.log("It is cloudy.");
        output.innerText += "It is cloudy.";
    }

    // ===== Single Random Age =====
    let age = Math.floor(Math.random() * 100) + 1; // Random age 1-100

    if (age < 18) {
        console.log(" You are " + age + " years old — a child.");
        output.innerText += " You are " + age + " years old — a child.";
    } else if (age <= 70) {
        console.log(" You are " + age + " years old — an adult.");
        output.innerText += " You are " + age + " years old — an adult.";
    } else { // age > 70
        console.log(" You are " + age + " years old — a senior.");
        output.innerText += " You are " + age + " years old — a senior.";
    }

})();
</script>
```

### <u>Popcorn Hack 2 - Vending Machine</u>

Instructions:
This hack is a bit more complicated than the first. Points will be awarded based on effort and completion.

Analyze the block of code below. Then, complete the following instructions:

You're hungry, so you go to your nearest vending machine for a snack. It makes you wonder if you'd be able to build your own. You can!... Digitally. Below, fill in the blanks so that when someone is prompted with food options, there are sub options within each food. (ex: Chocolate > Milk Chocolate)

1) **Create variables:**
- snackChoice -> user input (prompt) to determine which random snack is chosen (number from 1-3)

2) **Add basic conditionals determining what type of snack was chosen**

3) **Add nested if statements for extra behavior**

- Inside your chocolate condition:
    - Add user input to determine what type of chocolate the user wants
- Inside your candy condition:
    - Add user input to determine what color candy the user wants

4) **Challenge (OPTIONAL):**

- Make it so that there is a 10% chance of the vending machine giving you an extra cookie!

```python
%%html

<div id="outputPH2">Results will appear here:<br></div>

<script>
(() => {

    const output = document.getElementById("outputPH2");

    // 1. Prompt user to choose a snack
    let snackChoice = prompt("Choose snack 1, 2, or 3.");

    // 2. Determine snack choice
    if (snackChoice == "1") { // Chocolate bar
        let chocolateType = prompt("You got a chocolate bar. Choose your type: milk or dark");
        
        if (chocolateType) { // Only proceed if user entered something
            // Nested condition for chocolate type
            if (chocolateType.toLowerCase() === "milk") {
                output.innerText += "You chose a milk chocolate bar.";
            } else if (chocolateType.toLowerCase() === "dark") {
                output.innerText += "You chose a dark chocolate bar.";
            } else {
                output.innerText += "Unknown chocolate type. You still get a chocolate bar!";
            }
        }

    } else if (snackChoice == "2") { // Chips
        let chipsFlavor = prompt("You got a bag of chips. Choose a flavor: barbecue, or sour cream");
        
        if (chipsFlavor) { // Only proceed if user entered something
            if (chipsFlavor.toLowerCase() === "barbecue") {
                output.innerText += "You chose barbecue chips.";
            } else if (chipsFlavor.toLowerCase() === "sour cream") {
                output.innerText += "You chose sour cream chips.";
            } else {
                output.innerText += "Unknown flavor. You still get some chips!";
            }
        }

    } else if (snackChoice == "3") { // Candy
        let candyColor = prompt("You got candy. Choose the color: red, green, or yellow?");
        
        if (candyColor) { // Only proceed if user entered something
            // Nested condition for candy color
            if (candyColor.toLowerCase() === "red") {
                output.innerText += "You chose red candy.";
            } else if (candyColor.toLowerCase() === "green") {
                output.innerText += "You chose green candy.";
            } else if (candyColor.toLowerCase() === "yellow") {
                output.innerText += "You chose yellow candy.";
            } else {
                output.innerText += "Unknown candy color. You still get some candy!";
            }
        }

    } else {
        output.innerText += "Invalid snack choice. Please pick 1, 2, or 3.";
    }

})();
</script>
```

```python
%%html

<div id="outputHW">Results will appear here:</div>

<script>
(() => {

    const output = document.getElementById("outputHW");

    // 1) Temperature
    let temperature = 85; // You can change this
    if (temperature > 80) {
        console.log("It’s hot outside!");
        output.innerText += " It’s hot outside! ";
    } else {
        console.log("It’s nice outside.");
        output.innerText += " It’s nice outside. ";
    }

    // 2) Score with grading
    let score = Number(prompt("Enter your score (0-100):")); // Prompt for score

    if (score >= 90) {
        console.log("A");
        output.innerText += "A ";
    } else if (score >= 80) {
        console.log("B");
        output.innerText += "B ";
    } else if (score >= 70) {
        console.log("C");
        output.innerText += "C ";
    } else {
        console.log(" Needs improvement.");
        output.innerText += " Needs improvement. ";
    }

    // 3 & 4) Random number and even/odd check
    let number = Math.floor(Math.random() * 10) + 1; // Random integer 1-10
    if (number % 2 === 0) {
        console.log("Even Number");
        output.innerText += " Number " + number + ": Even Number.";
    } else {
        console.log("Odd Number");
        output.innerText += " Number " + number + ": Odd Number.";
    }

    // 5 & 6) Day switch statement and weekend check
    let day = "Saturday"; // Change to test
    switch(day) {
        case "Monday":
            console.log("Start of the week.");
            output.innerText += " Start of week.";
            break;
        case "Tuesday":
            console.log("It's Tuesday.");
            output.innerText += " It's Tuesday. ";
            break;
        case "Wednesday":
            console.log("It's wednesday.");
            output.innerText += " It's wednesday. ";
            break;
        case "Thursday":
            console.log("Almost Friday.");
            output.innerText += " Almost Friday. ";
            break;
        case "Friday":
            console.log("It's Friday!");
            output.innerText += " It's Friday! ";
            break;
        case "Saturday":
        case "Sunday":
            console.log("Weekend!");
            output.innerText += " Weekend! ";
            break;
        default:
            console.log("Unknown day.");
            output.innerText += "Unknown day. ";
    }

    // 7) Boolean loggedIn
    let loggedIn = true;
    if (loggedIn) {
        console.log("Welcome back!");
        output.innerText += " Welcome back! ";
    } else {
        console.log("Please log in.");
        output.innerText += " Please log in. ";
    }

    // 8) Weather
    let weather = "raining"; // change to test
    if (weather === "raining") {
        console.log("It's raining.");
        output.innerText += " It's raining. ";
    } else {
        console.log(" Go outside, it's a nice day today!");
        output.innerText += " Go outside, it's a nice day today! ";
    }

})();
</script>
```
